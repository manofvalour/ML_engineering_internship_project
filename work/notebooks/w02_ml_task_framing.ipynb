## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

The ML task type is Scoring(Ranking) problem. The objective is not to classify pages as "good" or "bad."

Instead, the model assigns each page an opportunity score that allows pages to be ranked from highest to lowest optimization potential.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The model predicts the page's expected click-through rate (CTR) given observable search characteristics.

the label can either be Expected CTR or Expected CTR - Actual CTR

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Model performance
  - RMSE or MAE (CTR prediction)

Business performance:
  - Percentage of recommended pages that truly underperform

Ultimately, success means that pages ranked highly by the model are worthwhile candidates for manual SEO review.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

Unit of Analysis
For my Lane's slice, One row == One page-query pair

Each observation corresponds to a single page appearing for a particular search query. a single content. A single row of the tabular data contain the session information for a content per user per day. And it includes the Google search console(GSC) and Google Analytics 4(GA4) information

Each row represents one page-query pair. The observed CTR corresponds to how users interacted with that page for that specific search query.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule treats every page the same, but expected CTR depends on many different factors. For example, a page in position 2 is expected to have a much higher CTR than a page in position 8, even if both are performing normally. Likewise, branded queries, informational queries, and transactional queries often have very different click behaviors. Search volume, page type, device, and SERP features can also influence CTR.

An if-statement such as "flag every page with CTR below 5%" ignores these interactions and would incorrectly flag many pages while missing others that are genuinely underperforming.
