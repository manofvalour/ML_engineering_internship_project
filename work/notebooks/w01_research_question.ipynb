## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

### What this work will be able to say:

  - This project provides observational, directional, and decision-support evidence.
  - It can identify pages that appear to receive fewer clicks than expected given their search position and other observable characteristics.
  - It can estimate which pages are likely to offer the greatest opportunity for improving click-through rate if reviewed.
  - It can rank pages by expected optimization value so SEO teams can prioritize limited time and resources.
  - It can quantify patterns and relationships between search visibility, page characteristics, and CTR within the available data.
  - It can support prioritization decisions with consistent, data-driven evidence rather than intuition or simple heuristics.


### What this work will never claim:**

  - This project does not prove causation or predict search engine behavior.
  - It does not prove that changing a page title, meta description, or content will increase CTR.
  - It does not claim to understand or reverse-engineer Google's ranking or ranking algorithms.
  - It does not predict future ranking positions or search engine updates.
  - It does not guarantee that a recommended page will generate additional traffic if optimized.
  - It cannot distinguish all external factors that influence CTR, such as competitors, SERP features, seasonality, or changes in user intent, unless those factors are explicitly represented in the data.

### Appropriate interpretation

The model should be viewed as a decision-support system, not an automated decision maker.
